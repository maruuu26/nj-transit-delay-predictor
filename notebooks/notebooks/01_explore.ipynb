```python
from pathlib import Path
import pandas as pd

print("Running from:", Path.cwd())

# walk upward from the current folder until we find data/archive
DATA = None
for folder in [Path.cwd(), *Path.cwd().parents]:
    if (folder / "data" / "archive").exists():
        DATA = folder / "data" / "archive"
        break

if DATA is None:
    raise FileNotFoundError("Couldn't find data/archive above this folder")

print("Found data at:", DATA)

df = pd.read_csv(DATA / "2018_03.csv")
df.head()
```
